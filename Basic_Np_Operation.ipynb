```python
import numpy as np
```

```python
salary = np.array([
    5000, 10000, 15000, 20000, 25000,
    30000, 35000, 40000, 45000, 50000,
    55000, 60000, 65000, 70000, 75000
])
print("Salary Data:", salary)
```

```python
# 2. operations
print("Mean Salary:", np.mean(salary))
print("Median Salary:", np.median(salary))
print("Average Salary:", np.average(salary))
print("Standard Deviation:", np.std(salary))
print("Variance:", np.var(salary))
print("Minimum Salary:", np.min(salary))
print("Maximum Salary:", np.max(salary))
print("Total Salary:", np.sum(salary))
```

```python
# 3. Filter salaries in above average
avg = np.mean(salary)

above_avg = salary[salary > avg]
print("Salaries Above Average:", above_avg)
```

```python
# 4. Filter salaries in  below average
below_avg = salary[salary < avg]
print("Salaries Below Average:", below_avg)
```

```python
# 5. Filter salaries in  specific range
range_salary = salary[(salary >= 40000) & (salary <= 75000)]
print("Salaries Between 40000 and 70000:", range_salary)
```

```python
# 6. Highest and Lowest Salary
print("Highest Salary:", np.max(salary))
print("Lowest Salary:", np.min(salary))
```

```python
# 7. Data Modification
salary[0] = 28000
print("Modified Salary Data:", salary)
```

```python

# 8. Sorting salaries
sorted_salary = np.sort(salary)
print("Sorted Salaries:", sorted_salary)
```
